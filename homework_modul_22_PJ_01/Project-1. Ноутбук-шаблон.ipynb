Нас будет интересовать только уровень образования.

Создайте с помощью функции-преобразования новый признак **"Образование"**, который должен иметь 4 категории: "высшее", "неоконченное высшее", "среднее специальное" и "среднее".

Выполните преобразование, ответьте на контрольные вопросы и удалите признак "Образование и ВУЗ".

Совет: обратите внимание на структуру текста в столбце **"Образование и ВУЗ"**. Гарантируется, что текущий уровень образования соискателя всегда находится в первых 2ух слов и начинается с заглавной буквы. Воспользуйтесь этим.

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:

* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
#ваш код здесь   
#функция преобразования
def experience_parse(experience):
    experience_itog = 0
    if pd.isnull(experience) or experience == 'Не указано':
        experience_itog = np.nan
        return experience_itog
    else :
        experience_arr = str(experience).split(' ')
        if experience_arr[3] in ('месяцев','месяца','месяц') :
            experience_itog = int(experience_arr[2])
            return experience_itog
        if experience_arr[3] in ('лет','года','год') and experience_arr[5] in ('месяцев','месяца','месяц'):
            experience_itog = (int(experience_arr[2])  * 12) + int(experience_arr[4])
            return experience_itog
        if experience_arr[3] in ('лет','года','год') and experience_arr[5] not in ('месяцев','месяца','месяц'):
            experience_itog = (int(experience_arr[2])  * 12)                          
        return experience_itog
#применение функции к новому столбцу
df_hh['Опыт работы (месяц)'] = df_hh['Опыт работы'].apply(experience_parse)
#удаление столбца
df_hh = df_hh.drop("Опыт работы", axis=1) 
```

Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
#ваш код здесь
million_cities = ['Новосибирск', 'Екатеринбург','Нижний Новгород','Казань', 'Челябинск','Омск', 
                  'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж','Волгоград']
#функция преобразования столбца город
def town_parse(town_in):
    town_itog = ''
    town_arr = str(town_in).split(' ,')
    if town_arr[0] in ('Москва','Санкт-Петербург'):
        town_itog = town_arr[0]
        return town_itog
    if town_arr[0] in million_cities:
        town_itog = 'город-миллионник'
        return town_itog
    else :
        town_itog = 'другие'
        return town_itog
#функция преобразования столбца Готовность к переезду    
def move_parse(move_in):
    move_itog = ''
    move_prepare = ''
    move_arr = str(move_in).split(' , ')
    if move_arr[1].find('м.') >= 0 :
        move_prepare = move_arr[2] 
    else :
        move_prepare = move_arr[1] 
    if move_prepare.find('не готов') >= 0:
        move_itog = False
    elif move_prepare.find('готов') >= 0 :
        move_itog = True
    elif move_prepare.find('хочу переехать') >= 0 :
        move_itog = True
    else :
        move_itog = np.nan
    return move_itog
#функция преобразования столбца Готовность к переезду  
def trip_parse(trip_in):
    if trip_in.find('не готов к командировкам') >= 0 :
        return False
    elif trip_in.find('не готова к командировкам') >= 0 :
        return False
    elif trip_in.find('готов к редким командировкам') >= 0 :
        return True   
    elif trip_in.find('готова к редким командировкам') >= 0 :
        return True 
    elif trip_in.find('готов к командировкам') >= 0 :
        return True
    elif trip_in.find('готова к командировкам') >= 0 :
        return True         
    else :
        return False

# #применение функции к новому столбцу
df_hh['Город'] = df_hh['Город, переезд, командировки'].apply(town_parse)
df_hh['Готовность к переезду'] = df_hh['Город, переезд, командировки'].apply(move_parse)
df_hh['Готовность к командировкам'] = df_hh['Город, переезд, командировки'].apply(trip_parse)
#удаление столбца
df_hh = df_hh.drop("Город, переезд, командировки", axis=1) 
# проверка
display(df_hh['Город'].unique())
display(df_hh['Готовность к переезду'].unique())
display(df_hh['Готовность к командировкам'].unique())

```

5. Рассмотрим поближе признаки **"Занятость"** и **"График"**. Сейчас признаки представляют собой набор категорий желаемой занятости (полная занятость, частичная занятость, проектная работа, волонтерство, стажировка) и желаемого графика работы (полный день, сменный график, гибкий график, удаленная работа, вахтовый метод).
На сайте hh.ru соискатель может указывать различные комбинации данных категорий, например:
* полная занятость, частичная занятость
* частичная занятость, проектная работа, волонтерство
* полный день, удаленная работа
* вахтовый метод, гибкий график, удаленная работа, полная занятость

Такой вариант признаков имеет множество различных комбинаций, а значит множество уникальных значений, что мешает анализу. Нужно это исправить!

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>
Выполните данное преобразование для признаков "Занятость" и "График", ответьте на контрольные вопросы, после чего удалите их из таблицы

```python
#ваш код здесь
#функция удаления лишних пробелов
def delete_space(str_in):
    job_itog = ''
    job_itog_arr = []
    job_split_arr = str(str_in).split(',')
    sep = ","
    job_itog_arr = [s.strip() for s in job_split_arr]
    job_itog = sep.join(job_itog_arr) 
    return job_itog
    
# Создание столбцов без лишних пробелов
df_hh['Занятость_итог'] = df_hh['Занятость'].apply(delete_space)
df_hh['График_итог'] = df_hh['График'].apply(delete_space)

#преобразование столбцов к виду One Hot Encoding
df_hh_job = df_hh['Занятость_итог'].str.get_dummies(sep=',').astype(bool)
df_hh_time = df_hh['График_итог'].str.get_dummies(sep=',').astype(bool)

#объединение таблиц
df_hh = pd.concat([df_hh,df_hh_job, df_hh_time], axis=1)

# ответы на вопросы в задании
display(df_hh[(df_hh['проектная работа'] == True) & (df_hh['волонтерство'] == True)].shape[0])
display(df_hh[(df_hh['вахтовый метод'] == True) & (df_hh['гибкий график'] == True)].shape[0])

# удаление лишних столбцов
df_hh.drop(['Занятость', 'График','Занятость_итог','График_итог'], axis=1, inplace=True)

```

3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
#ваш код здесь
df_currency = pd.read_csv('data/ExchangeRates.csv')
#Преобразование ДАТ к одному формату
df_hh['Обновление резюме'] = pd.to_datetime(df_hh['Обновление резюме'].astype("datetime64[ns]")).dt.date
df_currency['date'] = pd.to_datetime(df_currency['date'].astype("datetime64[ns]")).dt.date
# Преобразование столбца ЗП
df_hh[['wages','curname']] = df_hh['ЗП'].str.split(" ",expand=True)
#приведение к таблице соответсвтий
df_hh['curname'] = df_hh['curname'].apply(lambda x: str(x).replace('.', ''))
#Массив валют для приведения к ISO
new_cur = {
'oldname': ['грн', 'USD', 'EUR','белруб', 'KGS','сум','AZN','KZT','руб'],
'newname': ['UAH', 'USD', 'EUR','BYN', 'KGS','UZS','AZN','KZT','RUB'],
'ratio': [10, 1, 1, 1,10,10000,1,100,1]
}
df_new_cur = pd.DataFrame(new_cur)

#Объединение с таблицей соответствий
df_hh = df_hh.merge(df_new_cur, how='left', right_on='oldname', left_on='curname')

# Объединение с таблицей курсов
df_hh = df_hh.merge(df_currency[['currency', 'date','close']], 
                right_on=['currency', 'date'], left_on= ['newname','Обновление резюме'], 
                how='left')
#Замена для рублей по курсу 1
df_hh['close'] = df_hh['close'].fillna(1)

# Формирование итогового столбца ЗП(руб)
df_hh['ЗП (руб)'] = df_hh.apply(lambda row: (float(row['wages']) * float(row['close'])) / float(row['ratio']), axis=1)

# удаление лишних столбцов
df_hh.drop(['ЗП','wages', 'curname','oldname','newname','ratio','currency','close','date'], axis=1, inplace=True)
```

В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
# ваш код здесь
# Функция распределения
def age_z_score(data, feature, log_scale=False):
    if log_scale:
        x = np.log(data[feature])
    else:
        x = data[feature]
    mu = x.mean()
    sigma = x.std()
    lower_bound = mu - 3 * sigma
    upper_bound = mu + 4 * sigma
    outliers = data[(x < lower_bound) | (x > upper_bound)]
    cleaned = data[(x >= lower_bound) & (x <= upper_bound)]
    return outliers, cleaned

# Применям функцию распределения к возрасту
outliers, cleaned = age_z_score(df_hh, 'Возраст', log_scale=True)
print(f'Число выбросов по методу z-отклонения: {outliers.shape[0]}')

#гистограмма в логарифмическом масштабе
log_age = np.log(df_hh['Возраст'])
histplot = sns.histplot(log_age , bins=30)
histplot.axvline(log_age.mean(), color='k', lw=2)
histplot.axvline(log_age.mean()+ 4 * log_age.std(), color='k', ls='--', lw=2)
histplot.axvline(log_age.mean()- 3 * log_age.std(), color='k', ls='--', lw=2)
histplot.set(xlabel='Возраст', ylabel='Количество', title='Логарифмическое распределение возраста соискателей');
#Расчёт асимметрии 
display("Асимметрия: " + str(round(log_age.skew(),3)))
#таблица с выбросами
display(outliers)
# Удаляем строки
df_hh = df_hh.drop(index=outliers.index)
```
